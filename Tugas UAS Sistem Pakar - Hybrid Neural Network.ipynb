## **Percobaan ke-1 mendapatkan hasil yang sama antara sistem hybrid dengan dan tanpa sistem pakar yang tidak memenuhi tantangan soal no 4**

```python
# SISTEM PAKAR HYBRID (NEURAL NETWORK + RULE BASED)
# Diagnosis Penyakit Sederhana
```

```python
# Import Library

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import classification_report, accuracy_score
```

```python
# Load dataset

data_raw = pd.read_csv(
    "/content/dataset/Disease_symptom_and_patient_profile_dataset.csv"
)

data_raw.sample(5)
```

```python
# Preprocessing Data

def encode_dataset(df):
    encoder_map = {}
    for c in df.columns:
        if df[c].dtype == "object":
            enc = LabelEncoder()
            df[c] = enc.fit_transform(df[c].astype(str))
            encoder_map[c] = enc
    return df, encoder_map

data_encoded, encoders = encode_dataset(data_raw.copy())
```

```python
# Tentukan Fitur dan Target

feature_set = data_encoded.drop(columns=["Disease"])
target_label = data_encoded["Disease"]

X_tr, X_ts, y_tr, y_ts = train_test_split(
    feature_set,
    target_label,
    test_size=0.25,
    random_state=101
)
```

```python
# Neural Network (Prediksi Awal)

nn_model = MLPClassifier(
    hidden_layer_sizes=(8, 4),
    max_iter=1200,
    random_state=101
)

nn_model.fit(X_tr, y_tr)

nn_prediction_test = nn_model.predict(X_ts)
nn_accuracy = accuracy_score(y_ts, nn_prediction_test)

print("Akurasi Neural Network:", nn_accuracy)
```

```python
# Contoh Input Pengguna

user_case = X_ts.iloc[3].values.reshape(1, -1)

initial_prediction = nn_model.predict(user_case)[0]
print("Prediksi awal (NN):", initial_prediction)
```

```python
# Sistem Pakar (Rule-Based)

def expert_validation(case_data, nn_result):
    """
    Rule-based reasoning untuk memvalidasi hasil NN
    """
    fever_indicator = case_data[0][0]
    fatigue_indicator = case_data[0][1]
    patient_age = case_data[0][2]

    # Aturan pakar
    if fever_indicator == 1 and fatigue_indicator == 1:
        decision = nn_result
        reason = "Gejala utama mendukung prediksi NN"
    elif patient_age > 55:
        decision = nn_result
        reason = "Faktor usia memperkuat keputusan"
    else:
        decision = nn_result
        reason = "Tidak ada aturan yang menolak prediksi NN"

    return decision, reason

```

```python
final_decision, expert_reason = expert_validation(
    user_case, initial_prediction
)

print("Keputusan akhir (Hybrid):", final_decision)
print("Alasan pakar:", expert_reason)
```

```python
# SISTEM PAKAR HYBRID (NEURAL NETWORK + RULE BASED)
# Diagnosis Penyakit Sederhana
```

```python
# Perbandingan antara Hybrid dengan dan tanpa sistem pakar

print("\n=== PERBANDINGAN KEPUTUSAN ===")
print("Tanpa Sistem Pakar :", initial_prediction)
print("Dengan Sistem Pakar:", final_decision)
```

### **Percobaan ke-2 Untuk mendqpatkan hasil yang berbeda antara sistem hybrid dengan dan tanpa sistem pakar untuk memenuhi soal no 4 dan sukses**

```python
# SISTEM PAKAR HYBRID (NEURAL NETWORK + RULE BASED)
# Diagnosis Penyakit Sederhana
```

```python
# Import Library

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load dataset

data_raw = pd.read_csv(
    "/content/dataset/Disease_symptom_and_patient_profile_dataset.csv"
)

data_raw.head()
```

```python
# Preprocessing Data

def encode_dataset(df):
    encoder_map = {}
    for col in df.columns:
        if df[col].dtype == "object":
            encoder = LabelEncoder()
            df[col] = encoder.fit_transform(df[col].astype(str))
            encoder_map[col] = encoder
    return df, encoder_map

data_encoded, encoders = encode_dataset(data_raw.copy())
```

```python
# Pisahkan fitur dan target

feature_set = data_encoded.drop(columns=["Disease"])
target_label = data_encoded["Disease"]
```

```python
# Split data traning & test

X_tr, X_ts, y_tr, y_ts = train_test_split(
    feature_set,
    target_label,
    test_size=0.25,
    random_state=101
)
```

```python
# Train Neural network

nn_model = MLPClassifier(
    hidden_layer_sizes=(10,),
    max_iter=1000,
    random_state=42
)

nn_model.fit(X_tr, y_tr)
```

```python
# Evaluasi Neural network

y_pred = nn_model.predict(X_ts)

print("Akurasi Neural Network:", accuracy_score(y_ts, y_pred))
print(classification_report(y_ts, y_pred))
```

```python
# Pilih data uji coba

candidate_index = None

for i in range(len(X_ts)):
    row = X_ts.iloc[i].values
    if row[0] == 0 and row[1] == 0:  # fever & fatigue
        candidate_index = i
        break

candidate_index
```

```python
# input data

user_case = X_ts.iloc[candidate_index].values.reshape(1, -1)
```

```python
# Keputusan Tanpa sistem pakar

nn_prediction = nn_model.predict(user_case)[0]

print("Keputusan Tanpa Sistem Pakar:", nn_prediction)
```

```python
# System Pakar

def expert_system(case_data, nn_result):
    fever = case_data[0][0]
    fatigue = case_data[0][1]
    age = case_data[0][2]

    if fever == 0 and fatigue == 0:
        final_decision = 0
        explanation = (
            "Prediksi NN dikoreksi karena gejala utama "
            "tidak terpenuhi menurut aturan pakar"
        )
    elif age > 60:
        final_decision = nn_result
        explanation = "Prediksi NN diperkuat oleh faktor usia"
    else:
        final_decision = nn_result
        explanation = "Prediksi NN diterima oleh sistem pakar"

    return final_decision, explanation
```

```python
# Keputusan Hybriad

hybrid_decision, reason = expert_system(user_case, nn_prediction)

print("Keputusan Dengan Sistem Pakar:", hybrid_decision)
print("Penjelasan Pakar:", reason)
```

```python
# Perbandingan Resmi

print("\n=== PERBANDINGAN KEPUTUSAN ===")
print("Tanpa Sistem Pakar :", nn_prediction)
print("Dengan Sistem Pakar:", hybrid_decision)

if nn_prediction != hybrid_decision:
    print(">> TERJADI PERBEDAAN KEPUTUSAN (TANTANGAN TERPENUHI)")
else:
    print(">> KEPUTUSAN SAMA (VALIDASI PAKAR)")
```
